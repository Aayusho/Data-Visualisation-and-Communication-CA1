# Visualisation 3: Annual salary Distribution of Technology Occupations

```python
jobs = ['15-1252', '15-2051', '15-1221', '15-1212', '15-1211']

df[df['OCC_CODE'].isin(jobs)][
    ['OCC_CODE', 'OCC_TITLE', 'AREA_TYPE', 'AREA_TITLE', 'A_MEDIAN']
].head(15)
```

```python
df[['AREA_TYPE', 'AREA_TITLE']].drop_duplicates().head(20)
```

```python
box_data = df.loc[
    (df['OCC_CODE'].isin(jobs)) &
    (df['AREA_TYPE'] == 2) &
    (df['NAICS'] == '000000')
].copy()
```

```python
box_data['A_MEDIAN'] = pd.to_numeric(
    box_data['A_MEDIAN'],
    errors='coerce'
)

box_data = box_data.dropna(subset=['A_MEDIAN'])
```

```python
box_data.groupby('OCC_TITLE')['A_MEDIAN'].agg(
    ['count', 'min', 'median', 'max']
)
```

```python
jobs = ['15-1252', '15-2051', '15-1221', '15-1212', '15-1211']

salary_groups = [
    box_data.loc[
        box_data['OCC_CODE'] == code, 'A_MEDIAN'
    ].dropna().values
    for code in jobs
]
```

```python
labels = [
    'Software Developers',
    'Data Scientists',
    'Research Scientists',
    'Security Analysts',
    'Systems Analysts'
]

fig, ax = plt.subplots(figsize=(12, 6))

ax.boxplot(
    salary_groups,
    labels=labels,
    patch_artist=True
)

ax.set_title('Annual Salary Distribution of Technology Occupations (2023)')
ax.set_xlabel('Occupation')
ax.set_ylabel('Annual Median Salary (USD)')

plt.xticks(rotation=20, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### Visualisation 3: Explanation
The objective of this visualisation was to compare the distribution of annual median salaries 
accross USA for five technology related occupations. It helped identify salaries differncess and variances between these careers.

I chose box plot because it displays the median, spreads and possible outliers in the salary data. It allows us to compare how salaries vary accross USA for different type of occupations and identify which jobs have widely distributed salaries.

The box plot shows that Computer and information Research Scientists have the highest median at appox 120000 dollars, followed by Software Developers at around 17000 dollars, and Computer Systems Analysts have the lowest at 99000 dollars.
Data scientists show considerable salary variation while research scientists have a high outlier of approx 218000 dollars.

to sum it all up, the chart suggests that median salaries for technology occupations vary significantly across USA with some occupations showing greater salary differences than others.


# Visualisation 4: Employment vs Annual Median Salary

```python
scatter_codes = [
    '15-1252', '15-2051', '15-1221', '15-1212',
    '15-1211', '15-1242', '15-1241', '15-1253'
]

scatter_data = df.loc[
    (df['AREA'] == 99) &
    (df['NAICS'] == '000000') &
    (df['OCC_CODE'].isin(scatter_codes))
].copy()

scatter_data[['OCC_CODE', 'OCC_TITLE', 'TOT_EMP', 'A_MEDIAN']]
```

```python
scatter_data['TOT_EMP'] = pd.to_numeric(
    scatter_data['TOT_EMP'], errors='coerce'
)

scatter_data['A_MEDIAN'] = pd.to_numeric(
    scatter_data['A_MEDIAN'], errors='coerce'
)

scatter_data = scatter_data.dropna(
    subset=['TOT_EMP', 'A_MEDIAN']
)
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    scatter_data['TOT_EMP'],
    scatter_data['A_MEDIAN'],
    color='blue',
    s=100
)

plt.xlabel('Total Employment')
plt.ylabel('Annual Median Salary (USD)')
plt.title('Employment vs Annual Median Salary in Technology Occupations (2023)')

plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### Visualisation 4: Explanation
The objective of this visualisation was to investigate the relationship between total employment and annual median salaries across eight technology-related occupations in the USA in 2023. It helps us understand whether occupations with more employees also offer higher salaries.

I chose a scatter plot because it was useful for comparing two numerical variables and identifying possible relationships between them. Each dot represents a different occupation, making it easier to observe patterns, differences and unusual values in employment and salary.

The plot shows that software developers have the highest employment with approx 1.66 million workers and a median annual salary of 132000 dollars. As compare to the occupation with the highest median salary approx 145000 dollars has a much smaller workforce. 
Most occupations have fewer than 500000 employees with salaries ranging from approx 102000 dollars to 130000 dollars. 

The chart suggests that higher employment does not necesarily mean higher salaries. other factors such as skills, responsibilities and experience may influence earnings.

```python
df_2022[
    df_2022['OCC_TITLE'].str.contains(
        'Software Developers|Data Scientists|Computer and Information Research Scientists',
        case=False,
        na=False
    )
][['OCC_CODE', 'OCC_TITLE']].drop_duplicates()
```

```python
trend_codes = ['15-1252', '15-2051', '15-1221']
```

```python
trend_2021 = df_2021.loc[
    (df_2021['AREA'] == 99) &
    (df_2021['NAICS'] == '000000') &
    (df_2021['OCC_CODE'].isin(trend_codes))
].copy()

trend_2021[['OCC_TITLE', 'TOT_EMP']]
```

```python
trend_2022 = df_2022.loc[
    (df_2022['AREA'] == 99) &
    (df_2022['NAICS'] == '000000') &
    (df_2022['OCC_CODE'].isin(trend_codes))
].copy()

trend_2022[['OCC_TITLE', 'TOT_EMP']]
```

```python
trend_2023 = df.loc[
    (df['AREA'] == 99) &
    (df['NAICS'] == '000000') &
    (df['OCC_CODE'].isin(trend_codes))
].copy()

trend_2023[['OCC_TITLE', 'TOT_EMP']]
```

```python
trend_2021['Year'] = 2021
trend_2022['Year'] = 2022
trend_2023['Year'] = 2023
```

```python
trend_data = pd.concat(
    [trend_2021, trend_2022, trend_2023],
    ignore_index=True
)

trend_data[['Year', 'OCC_TITLE', 'TOT_EMP']]
```

```python
trend_data['TOT_EMP'] = pd.to_numeric(
    trend_data['TOT_EMP'],
    errors='coerce'
)
```

```python
employment_trends = trend_data.pivot(
    index='Year',
    columns='OCC_TITLE',
    values='TOT_EMP'
)

employment_trends
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

for occupation in employment_trends.columns:
    plt.plot(
        employment_trends.index,
        employment_trends[occupation],
        marker='o',
        linewidth=2,
        label=occupation
    )

plt.title('Employment Trends in Technology Occupations (2021–2023)')
plt.xlabel('Year')
plt.ylabel('Total Employment')
plt.xticks([2021, 2022, 2023])
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### Challenge task explanation
The objective of this visualisation was to analyse employment changes in Software Development, data science and computer research in the USA between 2021 and 2023.

I chose a line chart because it clearly displays changes over time. it allows employment trends accross different occupations to be compared and helps identify it. 

The data shows that employment increased across all three occupations between 2021 and 2023. Software Developers had the largest workforce, increasing from 1,364,180 to 1,656,880 employees.
Data Scientists experienced the highest percentage growth, increasing by approx 81.84%, while Software Developers grew by approx 21.46% and Computer and Information Research Scientists by approx 14.17%.

The results suggest that employment in these selected technology occupations increased during this period, with particularly strong growth in Data Science.
