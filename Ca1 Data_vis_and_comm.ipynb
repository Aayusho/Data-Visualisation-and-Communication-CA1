```python
codes = ['15-1221', '15-1252', '15-2051']

national_data = df.loc[
    (df['AREA'] == 99) &
    (df['NAICS'] == '000000') &
    (df['OCC_CODE'].isin(codes))
]

national_data[['OCC_CODE', 'OCC_TITLE', 'TOT_EMP']]
```

```python
national_data = national_data.copy()

national_data['TOT_EMP'] = pd.to_numeric(
    national_data['TOT_EMP'],
    errors='coerce'
)
```

```python
national_data[['OCC_CODE', 'OCC_TITLE', 'TOT_EMP']]
```

# Visualisation 1: Employment Comparision in AI-Related Occupation

```python

plt.figure(figsize=(10, 6))
plt.bar(national_data['OCC_TITLE'], national_data['TOT_EMP'])

plt.xlabel('Occupation')
plt.ylabel('Number of Employees')
plt.title('AI-Related Employment in the USA (2023)')

plt.xticks(rotation=20, ha='right')
plt.tight_layout()
plt.show()
```

### Visualisation 1 : Explaination
The objective of the visualisation is to compare the number of people employed in three technology related occupations in the USA in 2023 i.e. Software Developers, Data Scientists & Computer and Information Research Scientists.

I chose a bar chart because it is a simple and effective way to compare employment numbers between different occupations. It clearly shows which occupation has the highest and lowest employment.

The chart shows that Software Developers have the highest employment, with 1655880 workers followed by Data Scientists with 192710 workers. Computer and Information Research Scientists have the lowest employment with 35210 workers.
This chart helps us understand that Software Developers has a mucb larger workers compared to the other two. But these figures also do not indicate which occupations

# Visualisation 2 : Annual Median Salaries of AI-Related Occupations 

```python
salary_data = national_data.copy()

salary_data['A_MEDIAN'] = pd.to_numeric(
    salary_data['A_MEDIAN'],
    errors='coerce'
)

salary_data[['OCC_TITLE', 'A_MEDIAN']]
```

```python
plt.figure(figsize=(10, 5))

plt.scatter(
    salary_data['A_MEDIAN'],
    salary_data['OCC_TITLE'],
    color='blue',
    s=100
)

plt.xlabel('Annual Median Salary (USD)')
plt.ylabel('Occupation')
plt.title('Annual Median Salaries of AI-Related Occupations (2023)')

plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### Visualisation 2 : Explanation
The objective of this visualisation was to compare the annual median salaries of three technology-related occupations in the USA in 2023. It helped us understand the difference in earnings between these careers.

I chose a dot plot because it is a siple and effective way to compare numerical values accross different occupations. Each dot represents the median annual salary of an occupation making it easy to see salary differences. It also provides a clear visual comparison.

The plot shows that annual median salaries vary between the three occupations. it allows us to idfentify which occupation has the highest and lowest salary and compare the difference in earnings. there may be different factors affecting the differences so the chart alone cannot confirm the reason for that.
